```python
from sentence_transformers import SentenceTransformer
import numpy as np
from tqdm import tqdm

text_model = SentenceTransformer("Alibaba-NLP/gte-large-en-v1.5", trust_remote_code=True, token=token)


def encode_texts(df):
    texts = (df["Item Name"].fillna("") + " " + df["Description"].fillna("")).tolist()
    return np.array(text_model.encode(texts, batch_size=150, show_progress_bar=True))

# train_text_emb = encode_texts(train_df)
# val_text_emb = encode_texts(val_df)

# np.save("train_text_emb.npy", train_text_emb)
# np.save("val_text_emb.npy", val_text_emb)

test_text_emb = encode_texts(test_df)
np.save("test_text_emb.npy", test_text_emb)
```
