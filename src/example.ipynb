##  Read Dataset

```python
DATASET_FOLDER = '../dataset/'
train = pd.read_csv(os.path.join(DATASET_FOLDER, 'train.csv'))
test = pd.read_csv(os.path.join(DATASET_FOLDER, 'test.csv'))
sample_test = pd.read_csv(os.path.join(DATASET_FOLDER, 'sample_test.csv'))
sample_test_out = pd.read_csv(os.path.join(DATASET_FOLDER, 'sample_test_out.csv'))
```

```python
# find null rows form train set
train_null = train[train.isnull().any(axis=1)]
print(f"Number of rows with null values in train set: {len(train_null)}")

test_null = test[test.isnull().any(axis=1)]
print(f"Number of rows with null values in test set: {len(test_null)}")
```
