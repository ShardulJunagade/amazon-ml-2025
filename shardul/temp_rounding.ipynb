```python
import pandas as pd
```

```python
# i have a csv file wiht columns sample_id and price
df = pd.read_csv('submission.csv')
display(df.head())
```

```python
import numpy as np

# round to 3 decimals
df['price'] = df['price'].round(3)

# use boolean masks on the index to apply ceiling/floor every 10 rows
mask_ceil = df.index % 10 == 0
mask_floor = df.index % 10 == 5

df.loc[mask_ceil, 'price'] = np.ceil(df.loc[mask_ceil, 'price'])
df.loc[mask_floor, 'price'] = np.floor(df.loc[mask_floor, 'price'])

display(df.head(20))
```

```python
# save
df.to_csv('submission_rounded.csv', index=False)
```
