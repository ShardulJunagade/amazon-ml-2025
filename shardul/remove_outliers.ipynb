```python
import pandas as pd
import numpy as np

def remove_outliers_iqr(df, column, multiplier=1.5):
    """
    Remove outliers using the IQR (Interquartile Range) method.
    
    Parameters:
    - df: DataFrame
    - column: column name to check for outliers
    - multiplier: IQR multiplier (default 1.5, use 3.0 for extreme outliers only)
    
    Returns:
    - clean_df: DataFrame without outliers
    - outliers_df: DataFrame with only outliers
    """
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - multiplier * IQR
    upper_bound = Q3 + multiplier * IQR
    
    # Identify outliers
    outlier_mask = (df[column] < lower_bound) | (df[column] > upper_bound)
    
    clean_df = df[~outlier_mask]
    outliers_df = df[outlier_mask]
    
    print(f"Statistics for '{column}':")
    print(f"  Q1: {Q1:.2f}")
    print(f"  Q3: {Q3:.2f}")
    print(f"  IQR: {IQR:.2f}")
    print(f"  Lower bound: {lower_bound:.2f}")
    print(f"  Upper bound: {upper_bound:.2f}")
    print(f"\nOutliers found: {len(outliers_df)} ({len(outliers_df)/len(df)*100:.1f}%)")
    print(f"Clean data rows: {len(clean_df)} ({len(clean_df)/len(df)*100:.1f}%)")
    
    return clean_df, outliers_df


def remove_outliers_zscore(df, column, threshold=3):
    """
    Remove outliers using the Z-score method.
    
    Parameters:
    - df: DataFrame
    - column: column name to check for outliers
    - threshold: Z-score threshold (default 3)
    
    Returns:
    - clean_df: DataFrame without outliers
    - outliers_df: DataFrame with only outliers
    """
    mean = df[column].mean()
    std = df[column].std()
    
    z_scores = np.abs((df[column] - mean) / std)
    outlier_mask = z_scores > threshold
    
    clean_df = df[~outlier_mask]
    outliers_df = df[outlier_mask]
    
    print(f"Statistics for '{column}':")
    print(f"  Mean: {mean:.2f}")
    print(f"  Std Dev: {std:.2f}")
    print(f"  Threshold: {threshold} standard deviations")
    print(f"\nOutliers found: {len(outliers_df)} ({len(outliers_df)/len(df)*100:.1f}%)")
    print(f"Clean data rows: {len(clean_df)} ({len(clean_df)/len(df)*100:.1f}%)")
    
    return clean_df, outliers_df


def remove_outliers_percentile(df, column, lower_pct=1, upper_pct=99):
    """
    Remove outliers using percentile method.
    
    Parameters:
    - df: DataFrame
    - column: column name to check for outliers
    - lower_pct: lower percentile (default 1)
    - upper_pct: upper percentile (default 99)
    
    Returns:
    - clean_df: DataFrame without outliers
    - outliers_df: DataFrame with only outliers
    """
    lower_bound = df[column].quantile(lower_pct / 100)
    upper_bound = df[column].quantile(upper_pct / 100)
    
    outlier_mask = (df[column] < lower_bound) | (df[column] > upper_bound)
    
    clean_df = df[~outlier_mask]
    outliers_df = df[outlier_mask]
    
    print(f"Statistics for '{column}':")
    print(f"  {lower_pct}th percentile: {lower_bound:.2f}")
    print(f"  {upper_pct}th percentile: {upper_bound:.2f}")
    print(f"\nOutliers found: {len(outliers_df)} ({len(outliers_df)/len(df)*100:.1f}%)")
    print(f"Clean data rows: {len(clean_df)} ({len(clean_df)/len(df)*100:.1f}%)")
    
    return clean_df, outliers_df


# Main execution
# Read the CSV file
# input_file = "your_data.csv"  # Change this to your file name
# df = pd.read_csv(input_file)

print(f"Original data: {len(df)} rows\n")
print(f"Price statistics:")
print(df['price'].describe())
print("\n" + "="*60 + "\n")

# # Method 1: IQR method (RECOMMENDED)
# print("Method 1: IQR (Interquartile Range)")
# print("-" * 60)
# clean_df, outliers_df = remove_outliers_iqr(df, 'price', multiplier=1.5)

# # Save cleaned data
# clean_df.to_csv('cleaned_data.csv', index=False)
# outliers_df.to_csv('outliers.csv', index=False)

# print("\n✓ Files saved:")
# print("  - cleaned_data.csv")
# print("  - outliers.csv")

# Uncomment to try other methods:

# Method 2: Z-score method
# print("\n\nMethod 2: Z-Score")
# print("-" * 60)
# clean_df, outliers_df = remove_outliers_zscore(df, 'price', threshold=3)

# Method 3: Percentile method
# print("\n\nMethod 3: Percentile")
# print("-" * 60)
# clean_df, outliers_df = remove_outliers_percentile(df, 'price', lower_pct=1, upper_pct=99)
```
